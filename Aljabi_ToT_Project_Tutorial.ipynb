## Shell Scripting 
* **Shell Scripting** is a list of commands controlling programs and their flow. It is run by a **Unix shell**, which is a command line interpreter. 
* When a program runs, which programs communicate with each other, and how these programs process data are main actions that can be achieved using shell scripting along with file manipulation and text printing. 
* There's also a **Wrapper**, which is a kind of shell script that created the programming environment, runs the progrma, etc.
* Understanding these terms is important but the main point to understand is that shell scripting is performing a list of commands through the terminal window.
* These commands can range from printing a simple sentence to making an entire program that runs on percise daily basis. It is unreal the amount of control and power shell script gives the user.
* This tool allows you to take control of your machine on a deeper level. You can pretty much control every part of your computer and give percise commands to follow.

### Permissions
* Every file has three permissions:
    * owner-group-others
* We use `chmod` and the permission amount after.    
* The lay out:
        0 = 0 = nothing
        1 = 1 = execute x
        2 = 2 = write w
        3 = 2+1 = w+x
        4 = 4 = read r
        5 = 4+1 = r+x
        6 = 4+2 = r+w
        7 = 4+2+1 = r+w+x
* To give a file permission you need to specify it for every one. So full permission would look like this:

        chmod 777 example.txt
        
* To use a script, we need to give it permission to execute first which is why understanding the grid above is important.

### operators
* There are a bunch of important operators shell scripters use in almost every shell. 
    * Relational operators:
        * `-eq` : equal to
        * `-ne` : not equal to
        * `-gt` : greater than
        * `-lt` : less than
        * `-ge` : greater than or equal
        * `-le` : less than or equal
    * Boolean operators:
        * `!` : logical NOT
        * `-o` : logical OR
        * `-a` : logica AND
    * String operators:
        * `=` : equal
        * `!=` : not equal
        * `-z` : checks length (True if 0)
        * `-n` : checks length (True if not 0)
        * `str` : checks if empty string (True if not empty)
    * File test operators:
        * `-d` : checks if the file is a directory
        * `-f` : checks if it's an ordinary file
        * `-c` : checls if it's a special file
        * `-r` : checks is readable
        * `-s` : checks if file is not empty
        * `-w` : checls if file can be written to and modified
        * `-x` : checks if file can be executed
* There are more operators but these are the ones that are used the most, and the ones new shell scripters need to understand.

        

### Variables
* Assigning variables in a script is very similar to other programming languages. The syntax:
        Variable=Value
* **Scaler variables** is what we call the normal variables. 
* During assignment, you can't put a space after or before the equal sign.
* When using the variable in a command, you put a dollar sign behind it.
* Other types of variables are the **readonly** and **unset** variables. 
    * **readonly variable** is a variable that cannot be changed or modified.
    * **unset variable** is a variable that is removed from the variable list.
* **Special variables** are variables with special functions. They start with a dollar sign and each have a different use.
    * `$#` : no. of parameters
    * `$*` : all quoted values
    * `$@` : individual auoted values
    * `$?` : exit status or return value.(0 if succesful, 1 if unsuccesful)
    * `$$` : process number/ id number of the shell
    * `$1,2,....,9` : argument parameter
    * `$0` : file name of script
